```python
import kagglehub
import shutil
import os
import pandas as pd
import glob

# Step 1: Download dataset
path = kagglehub.dataset_download("emineyetm/fake-news-detection-datasets")

# Step 2: Copy to current directory
dest = os.path.join(os.getcwd(), "fake_news_dataset")
shutil.copytree(path, dest, dirs_exist_ok=True)

print("Dataset copied to:", dest)

# Step 3: Collect all possible data files
csv_files = glob.glob(f"{dest}/**/*.csv", recursive=True)
excel_files = glob.glob(f"{dest}/**/*.xlsx", recursive=True)
json_files = glob.glob(f"{dest}/**/*.json", recursive=True)
txt_files = glob.glob(f"{dest}/**/*.txt", recursive=True)

all_dfs = []

# Step 4: Read CSV
for file in csv_files:
    try:
        df = pd.read_csv(file)
        all_dfs.append(df)
        print("Loaded CSV:", file)
    except:
        print("Skipped CSV:", file)

# Step 5: Read Excel
for file in excel_files:
    try:
        df = pd.read_excel(file)
        all_dfs.append(df)
        print("Loaded Excel:", file)
    except:
        print("Skipped Excel:", file)

# Step 6: Read JSON
for file in json_files:
    try:
        df = pd.read_json(file)
        all_dfs.append(df)
        print("Loaded JSON:", file)
    except:
        print("Skipped JSON:", file)

# Step 7: Read TXT (try tab-separated)
for file in txt_files:
    try:
        df = pd.read_csv(file, sep="\t")
        all_dfs.append(df)
        print("Loaded TXT:", file)
    except:
        print("Skipped TXT:", file)

# Step 8: Merge everything
if all_dfs:
    final_df = pd.concat(all_dfs, ignore_index=True)
    final_df.to_csv("final_dataset.csv", index=False)
    print("✅ Final CSV saved as: final_dataset.csv")
else:
    print("❌ No usable files found")
```

```python
import pandas as pd

df = pd.read_csv("final_dataset.csv")

print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
print(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```
